# Construction Project Risk Classification Using Machine Learning

## A Comparative Study of Logistic Regression, Decision Tree, and Random Forest Models

### 1. Project Overview

Construction projects are exposed to multiple sources of uncertainty, including cost overruns, schedule deviations, safety incidents, structural conditions, environmental factors, resource utilization, and project execution issues. Early identification of project risk can support construction managers, engineers, and project stakeholders in prioritizing monitoring, allocating resources, and implementing corrective actions.

This project develops a supervised machine learning classification system for predicting the risk level of civil engineering and construction projects using project, cost, schedule, structural, environmental, resource, safety, and monitoring-related data.

The project uses a BIM and AI-oriented civil engineering dataset containing project-level observations and a categorical risk outcome. The target variable, `Risk_Level`, contains three classes:

- **Low Risk**
- **Medium Risk**
- **High Risk**

Rather than reducing the problem to binary classification, this project treats risk assessment as a **multiclass classification problem**, preserving the three levels of risk represented in the original dataset.

---

## 2. Business and Construction Problem

Construction managers need to identify projects that may require different levels of management attention.

A practical risk classification framework can be interpreted as:

- **Low Risk:** Normal project monitoring and routine management.
- **Medium Risk:** Increased monitoring and potential corrective intervention.
- **High Risk:** Priority risk assessment and immediate management attention.

The central question of this project is:

> **Can machine learning models accurately classify construction projects into Low, Medium, and High risk categories using available project-related information?**

The project also investigates which classification algorithm provides the strongest predictive performance and what the results may imply from a construction project-management perspective.

## 5. Machine Learning Formulation

This project formulates construction risk assessment as a **multiclass supervised classification problem**.

The target classes are:

| Class | Meaning |
|---|---|
| Low | Lower project risk |
| Medium | Moderate project risk |
| High | Higher project risk |

The models will learn patterns in the predictor variables and attempt to classify previously unseen projects into one of these three risk categories.

---

## 6. Models

Three classification algorithms will be developed and compared:

### Logistic Regression

Used as a relatively interpretable baseline classification model. It provides a useful benchmark against which the more flexible tree-based models can be evaluated.

### Decision Tree

A nonlinear model capable of learning decision rules and interactions between construction project variables.

### Random Forest

An ensemble learning method that combines multiple decision trees and may provide stronger predictive performance and improved robustness compared with a single decision tree.

---

## 7. Evaluation Framework

The models will be evaluated using multiple performance metrics rather than relying solely on accuracy.

The evaluation will include:

- Accuracy
- Precision
- Recall
- F1-score
- Confusion matrix
- Macro-averaged metrics
- Weighted-averaged metrics
- ROC-AUC where appropriate

The confusion matrix will be particularly important because it will show how often the models confuse one risk category with another.

For example, incorrectly classifying a genuinely **High Risk** project as **Low Risk** may have greater practical implications than classifying a High Risk project as Medium Risk.

Therefore, model evaluation will consider both statistical performance and the potential construction-management consequences of classification errors.

---

## 8. Data Leakage and Model Validity

A major focus of this project is preventing **data leakage**.

Some variables in a construction risk dataset may be closely related to, or potentially used in the construction of, the `Risk_Level` target. If such variables are supplied directly to a model, the resulting performance may be artificially inflated and may not represent genuine predictive capability.

Therefore, the project will investigate the relationship between the target and potential predictor variables before model training.

Features will be selected based on their appropriateness for the intended prediction task rather than simply using every available column.

# 6. Target Leakage Investigation

Target leakage occurs when information that would not legitimately be available at prediction time, or information directly used to construct the target, is included among the model predictors.

Leakage can produce artificially high model performance and make a machine learning system appear more capable than it actually is.

Because the target variable in this dataset is `Risk_Level`, variables that are highly associated with the target will be investigated before model training.

Particular attention will be given to variables such as:

- `Safety_Risk_Score`
- `Cost_Overrun`
- `Schedule_Deviation`
- `Anomaly_Detected`
- `Completion_Percentage`

The purpose is not to remove variables merely because they correlate with risk. Instead, the project will determine whether each variable represents a legitimate predictor or a component/derived representation of the target itself.

## 7.4 Feature Selection Decisions

Based on the exploratory data analysis and domain reasoning, the following feature-selection decisions were made:

### Identifier
`Project_ID` will be excluded because it uniquely identifies each project but does not represent a meaningful construction characteristic. Including it could cause the model to learn project-specific identifiers rather than generalizable patterns.

### Target Variable
`Risk_Level` is the target variable to be predicted. It contains three classes:

- Low
- Medium
- High

It will therefore be excluded from the predictor matrix.

### Categorical Predictors
The following categorical variables will be retained:

- `Project_Type`
- `Location`
- `Weather_Condition`

These variables represent potentially meaningful construction and environmental factors.

### Numerical Predictors
Numerical project-performance and risk indicators will be evaluated as candidate predictors. `Safety_Risk_Score` will be retained because it demonstrated a strong relationship with project risk while showing overlapping distributions across the risk classes.

The final predictor set will be validated during model development to identify potential redundancy, leakage, or low-value variables.

# 📊 Feature Importance Analysis (XGBoost)

## Overview

Feature importance analysis helps understand which input variables contributed the most to the XGBoost model's predictions. Unlike simple statistical correlations, XGBoost evaluates how much each feature reduces prediction error while constructing the ensemble of decision trees.

Understanding feature importance improves the interpretability of the model and provides valuable insights into the key factors influencing construction project risk.

---

## Key Findings

The feature importance visualization indicates that the model relied most heavily on operational, project-specific, and environmental variables when predicting project risk levels.

The most influential features include:

- Anomaly_Detected
- Project Location
- Project Type
- Calculated_Duration_Days
- Equipment_Utilization
- Labor_Hours
- Planned_Cost
- Planned_Duration
- Crack_Width
- Humidity

---
